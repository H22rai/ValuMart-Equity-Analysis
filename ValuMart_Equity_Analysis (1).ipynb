## Analysis

```python
pay_by_dept_gender = dtEmployees.groupby(['Department', 'Gender']).agg(
    avgHourlyRate=('HourlyRate', 'mean'),
    employeeCount=('EmployeeID', 'count')
).unstack()
pay_by_dept_gender
```

```python
pay_gap = dtEmployees.groupby(['Department', 'Gender'])['HourlyRate'].mean().unstack()
pay_gap['payGapPct'] = (
    (pay_gap['M'] - pay_gap['F']) / pay_gap['M'] * 100
).round(2)
pay_gap
```

```python
threshold = 5.0
flagged_depts = pay_gap.loc[pay_gap['payGapPct'] > threshold].copy()
print(f'Departments with pay gap exceeding {threshold}% threshold:')
display(flagged_depts)
print(f'\nNumber of departments breaching threshold: {len(flagged_depts)}')
```

```python
dtEmployees.groupby('Department')['YearsTenure'].mean().sort_values(ascending=False)
```

```python
dtEmployees.groupby(['StoreID','Gender'])['EmployeeID'].count().head(10)
```

```python
dtEmployees.merge(dtStores[['StoreID','Region']], on='StoreID').groupby('Region')['HourlyRate'].mean()
```

```python
dtEmployees.loc[dtEmployees['Department'] == 'Management'].sort_values('HourlyRate', ascending=False).head(10)
```

```python
affected = dtEmployees.loc[
    (dtEmployees['Department'].isin(flagged_depts.index)) &
    (dtEmployees['Gender'] == 'F')
]
print(f'Female employees in flagged departments: {len(affected)}')

male_avg = dtEmployees.loc[
    dtEmployees['Department'].isin(flagged_depts.index) &
    (dtEmployees['Gender'] == 'M')
].groupby('Department')['HourlyRate'].mean()

affected = affected.merge(male_avg.rename('maleAvgRate'), on='Department', how='left')
affected['annualPayGap'] = (
    (affected['maleAvgRate'] - affected['HourlyRate']) *
    affected['ContractedWeeklyHours'] * 52
)
total_gap = affected['annualPayGap'].sum()
print(f'Total annual pay gap exposure: ${total_gap:,.2f}')
```
